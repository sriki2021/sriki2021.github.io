**Black holes** are enormous astronomical bodies, with all their mass compressed into an tiny point known as the **singularity**. Their gravitational pull on nearby objects is so strong that even if light rays come close enough, (cross a boundary known as the **event horizon**) they cannot escape. They have a multitude of other interesting characteristics (time dilation, lensing of light, and potentially even portals to other universes!) that make them worth exploring. Considering their extreme properties, scientists also use them to test possible theories at the fringes of what we know definitively, to try and bridge the gap between **general relativity** (the theory describing black holes and our universe at a large scale) and **quantum mechanics** (the theory describing very small matter, such as in the singularity of a black hole).
